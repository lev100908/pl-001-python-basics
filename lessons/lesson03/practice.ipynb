## 1. Функции

До сих пор мы уже пользовались функциями — `print()`, `input()`, `abs()`,
`round()`. Пока что достаточно временного упрощенного представления:
функция — это просто **именованный блок кода**, который можно
переиспользовать много раз, не переписывая его заново. Позже в курсе
разберем функции подробнее — с более точным и полным определением.

### Определение функции

```python
def имя_функции(параметр1, параметр2):
    тело_функции
    return результат
```

- `def` — ключевое слово, с которого начинается определение функции;
- в скобках перечисляются **параметры** — имена, через которые в функцию
  попадают входные данные;
- **тело функции** — идущий после строки с `def` набор инструкций с
  одинаковым уровнем отступа; эти инструкции выполняются каждый раз, когда
  функция вызывается. Внутри тела можно использовать параметры функции —
  так же, как обычные переменные;
- `return` — необязательное ключевое слово, которое завершает выполнение
  функции и возвращает результат туда, откуда функция была вызвана.
### Вызов функции

После того как функция определена, ее можно вызывать по имени, передавая
на место параметров конкретные значения — **аргументы**:

```python
имя_функции(аргумент1, аргумент2)
```

````python
def square(number: int) -> int:
    return number ** 2


print(square(4))    # 16
print(square(-3))   # 9

result = square(5)  # результат можно сохранить в переменную
print(result)       # 25
````

### Параметры со значением по умолчанию

Параметру можно задать значение **по умолчанию** — тогда при вызове
соответствующий аргумент можно не указывать, и будет использовано значение
по умолчанию:

````python
def greet(name: str, greeting: str = "Hello") -> str:
    return f"{greeting}, {name}!"


print(greet("Anna"))         # Hello, Anna!
print(greet("Anna", "Hi"))   # Hi, Anna!
````

## 2. Задание: наибольший общий делитель

Наибольший общий делитель (НОД) двух целых чисел — самое большое число, на
которое оба числа делятся без остатка. Один из самых старых известных
алгоритмов для его вычисления — **алгоритм Евклида**, основанный на двух
фактах:

- `gcd(a, 0) == a`;
- `gcd(a, b) == gcd(b, a % b)`, если `b != 0`.

То есть можно взять остаток от деления `a` на `b`, заменить им `b`, а
бывшее `b` сделать новым `a` — и повторять это, пока второе число не
станет равным `0`. Тогда первое число и будет ответом.

Реализуйте функцию `gcd(first_number, second_number)`, которая вычисляет
наибольший общий делитель `first_number` и `second_number` с помощью
цикла `while` по описанному алгоритму.

````python
def gcd(first_number: int, second_number: int) -> int:
    if second_number == 0:
        return first_number
    else:
        return gcd(second_number, first_number % second_number)
````

### Проверка

````python
assert gcd(48, 18) == 6
assert gcd(17, 5) == 1
assert gcd(0, 5) == 5
assert gcd(7, 7) == 7

print("OK")
````

## 4. Встроенная библиотека `math`

В Python уже есть встроенная библиотека `math` с функциями для
математических вычислений — не нужно писать самостоятельно то, что уже
реализовано (и хорошо протестировано) в стандартной библиотеке. Чтобы
пользоваться библиотекой, ее нужно импортировать: `import math`, а дальше
обращаться к ее функциям и константам через точку: `math.имя`.

### `math.isclose` — то же самое, что мы только что написали сами

В `math` уже есть готовая функция для сравнения чисел с плавающей точкой —
`math.isclose(a, b, rel_tol=1e-09, abs_tol=0.0)`. По умолчанию она
сравнивает числа с **относительной** погрешностью `rel_tol` (то есть
допустимая разница зависит от величины самих чисел), а `abs_tol` задает
**абсолютную** погрешность — то, что мы реализовали в `is_close` выше.

````python
import math

print(math.isclose(0.1 + 0.2, 0.3))                 # True
print(math.isclose(1.0, 1.0001))                    # False
print(math.isclose(1.0, 1.0001, rel_tol=1e-3))      # True
````

### Другие возможности `math`

Некоторые часто используемые функции и константы `math`: `math.sqrt` —
квадратный корень, `math.floor` — округление вниз, `math.ceil` —
округление вверх, `math.trunc` — отбрасывание дробной части,
`math.factorial` — факториал, `math.gcd` — наибольший общий делитель (то,
что мы писали сами!), `math.pi` и `math.e` — константы, `math.sin` —
синус, `math.log` — натуральный логарифм.

Это далеко не все — в `math` десятки функций (тригонометрия, логарифмы,
гиперболические функции, специальные функции и др.). Со всем набором
можно ознакомиться в официальной документации:
[Python documentation — math](https://docs.python.org/3/library/math.html).

````python
print(math.sqrt(16))          # 4.0
print(math.floor(3.7))        # 3
print(math.ceil(3.2))         # 4
print(math.trunc(-3.7))       # -3
print(math.factorial(5))      # 120
print(math.gcd(48, 18))       # 6
print(math.pi)                # 3.141592653589793
print(math.e)                 # 2.718281828459045
print(math.sin(math.pi / 2))  # 1.0
print(math.log(math.e))       # 1.0
````

## 5. Встроенная библиотека `decimal`

Если в задаче важна именно **десятичная точность** (например, денежные
расчеты, где недопустимы даже мелкие погрешности `float`), вместо `float`
стоит использовать тип `Decimal` из встроенной библиотеки `decimal`. В
отличие от `float`, `Decimal` хранит число в десятичном виде и не
подвержен погрешностям двоичного представления.

````python
from decimal import Decimal

print(0.1 + 0.2)                          # 0.30000000000000004 -- погрешность float
print(Decimal("0.1") + Decimal("0.2"))    # 0.3                 -- точно!
````

> **Осторожно.** `Decimal` нужно создавать из строки (`Decimal("0.1")`), а
> не из готового `float` (`Decimal(0.1)`) — иначе в `Decimal` попадет уже
> неточное значение, которое `float` успел исказить при хранении:

```python
Decimal(0.1)     # 0.1000000000000000055511151231257827021181583404541015625
Decimal("0.1")   # 0.1 -- правильный способ
```

`Decimal` поддерживает все обычные арифметические операции (`+ - * / **`),
а точность вычислений (число значащих цифр) можно настраивать через
`decimal.getcontext().prec`. Подробнее в
[официальной документации](https://docs.python.org/3/library/decimal.html).

## 6. Битовые операции

На лекции мы разобрали основные операции над числами, но не рассмотрели
еще один их набор — **битовые операции**. Они эксклюзивны для целых чисел
(`int`): в отличие от арифметических операций, для `float` они не
определены и приведут к ошибке.

Мы уже знаем, что целое число можно записать в двоичном виде (литерал
`0b...`, функция `bin()`). Битовые операции позволяют работать с числом
именно на уровне отдельных битов его двоичного представления — это часто
применяется при работе с флагами, масками, кодировании данных и в задачах,
где важна эффективность по памяти или скорости.

| Оператор | Название | Что делает |
|---|---|---|
| `~x` | побитовое НЕ | инвертирует каждый бит числа `x` |
| `x & y` | побитовое И | бит результата — `1`, если оба соответствующих бита `x` и `y` — `1` |
| `x \| y` | побитовое ИЛИ | бит результата — `1`, если хотя бы один из соответствующих битов `x` или `y` — `1` |
| `x ^ y` | побитовое исключающее ИЛИ | бит результата — `1`, если соответствующие биты `x` и `y` различаются |
| `x << n` | сдвиг влево | сдвигает биты `x` влево на `n` позиций (эквивалентно `x * 2 ** n`) |
| `x >> n` | сдвиг вправо | сдвигает биты `x` вправо на `n` позиций (эквивалентно `x // 2 ** n`) |

`~` — унарная операция (один операнд), все остальные — бинарные (два
операнда).

> **Осторожно.** В Python `int` не хранится в фиксированном количестве
> бит (в отличие, например, от C), поэтому побитовое НЕ (`~x`) — это не
> "переворот" видимых битов, а `-(x + 1)`, что следует из представления
> отрицательных чисел в дополнительном коде. Поэтому `~5` дает `-6`, а не
> какое-то "перевернутое" положительное число.

````python
first_number = 13    # 0b01101
second_number = 6    # 0b00110

print(f"first_number  = 0b{first_number:05b} = {first_number:3d}")
print(f"second_number = 0b{second_number:05b} = {second_number:3d}")
print()
print(
    f"~first_number                = {~first_number:4d}  # bitwise NOT: -(x + 1)"
)
print(
    f"first_number & second_number = 0b{first_number & second_number:05b}"
    f" = {first_number & second_number:3d}  # AND"
)
print(
    f"first_number | second_number = 0b{first_number | second_number:05b}"
    f" = {first_number | second_number:3d}  # OR"
)
print(
    f"first_number ^ second_number = 0b{first_number ^ second_number:05b}"
    f" = {first_number ^ second_number:3d}  # XOR"
)
print(
    f"first_number << 1            = 0b{first_number << 1:05b}"
    f" = {first_number << 1:3d}  # left shift"
)
print(
    f"first_number >> 1            = 0b{first_number >> 1:05b}"
    f" = {first_number >> 1:3d}  # right shift"
)
````
