## 3. Операции теории множеств

Множества в Python поддерживают операции из теории множеств. У каждой
операции есть две формы: **оператор** и **метод**.

| Операция | Оператор | Метод | Результат |
|---|---|---|---|
| Объединение | `a \| b` | `a.union(b)` | элементы, которые есть хотя бы в одном множестве |
| Пересечение | `a & b` | `a.intersection(b)` | элементы, которые есть в обоих множествах |
| Разность | `a - b` | `a.difference(b)` | элементы `a`, которых нет в `b` |
| Симметрическая разность | `a ^ b` | `a.symmetric_difference(b)` | элементы, которые есть ровно в одном из множеств |

Все эти операции **не изменяют** исходные множества, а возвращают новое.

Результат каждой операции ниже показан на **диаграмме Эйлера**: множество
`A` — это `first_set`, множество `B` — `second_set`, а синим закрашена область,
которая попадает в результат.

Возьмем два множества для экспериментов. Выводить будем через `sorted()`,
чтобы порядок элементов был предсказуемым:

```python
first_set = {1, 2, 3, 4, 5}
second_set = {4, 5, 6, 7}

print("first_set: ", sorted(first_set))
print("second_set:", sorted(second_set))
```

### Объединение

![Объединение A ∪ B](images/union.svg)

```python
print("operator:", sorted(first_set | second_set))
print("method:  ", sorted(first_set.union(second_set)))
```

### Пересечение

![Пересечение A ∩ B](images/intersection.svg)

```python
print("operator:", sorted(first_set & second_set))
print("method:  ", sorted(first_set.intersection(second_set)))
```

### Разность

Разность **не коммутативна**: порядок операндов важен.

![Разность A − B](images/difference.svg)
![Разность B − A](images/difference_reversed.svg)

```python
print("first - second")
print("  operator:", sorted(first_set - second_set))
print("  method:  ", sorted(first_set.difference(second_set)))

print("second - first")
print("  operator:", sorted(second_set - first_set))
print("  method:  ", sorted(second_set.difference(first_set)))
```

### Симметрическая разность

Это «объединение без пересечения»: элементы, которые есть ровно в одном из
двух множеств. Операция коммутативна.

![Симметрическая разность A Δ B](images/symmetric_difference.svg)

```python
print("operator:", sorted(first_set ^ second_set))
print("method:  ", sorted(first_set.symmetric_difference(second_set)))
```

### Чем отличаются операторы и методы

1. **Операторы** требуют, чтобы оба операнда были множествами. **Методы**
   принимают любой итерируемый объект (список, кортеж, строку и т. д.):

```python
print(sorted(first_set.union([10, 11], (12,))))
print(sorted(first_set.intersection([1, 2, 99])))
print(sorted({"a", "b", "c"}.difference("bx")))
```

```python
first_set | [10, 11]
```

2. Методы `union()` и `intersection()` принимают **несколько аргументов**
   сразу:

```python
third_set = {5, 7, 9}

print("union:       ", sorted(first_set.union(second_set, third_set)))
print("intersection:", sorted(first_set.intersection(second_set, third_set)))
```

Исходные множества при этом не изменились:

```python
print(sorted(first_set), sorted(second_set))
```

### Подмножество и надмножество

Множество `a` называется **подмножеством** множества `b`, если каждый
элемент `a` есть и в `b`. В этом случае `b` называют **надмножеством** `a`.
Для проверки есть методы и операторы сравнения:

| Отношение | Оператор | Метод | Результат `True`, если |
|---|---|---|---|
| Подмножество | `a <= b` | `a.issubset(b)` | все элементы `a` есть в `b` (множества могут совпадать) |
| Строгое подмножество | `a < b` | — | `a` — подмножество `b` и `a != b` |
| Надмножество | `a >= b` | `a.issuperset(b)` | все элементы `b` есть в `a` (множества могут совпадать) |
| Строгое надмножество | `a > b` | — | `a` — надмножество `b` и `a != b` |

У строгих сравнений `<` и `>` методов-аналогов нет. Пустое множество —
подмножество любого множества, а любое множество — подмножество самого себя.

![Подмножество A ⊂ B](images/subset.svg)

```python
small_set = {1, 2}
big_set = {1, 2, 3, 4}

print("small <= big:            ", small_set <= big_set)
print("small.issubset(big):     ", small_set.issubset(big_set))
print("big >= small:            ", big_set >= small_set)
print("big.issuperset(small):   ", big_set.issuperset(small_set))
print("big <= small:            ", big_set <= small_set)
```

Нестрогие сравнения `<=` и `>=` допускают равенство множеств, а строгие `<` и
`>` — нет: множество не является **строгим** подмножеством самого себя.

```python
same_set = {1, 2}

print("small <= same:", small_set <= same_set)
print("small <  same:", small_set < same_set)
print("small <  big: ", small_set < big_set)
print("empty <= big: ", set() <= big_set)
```

Как и в случае с другими операциями, **методы** принимают любой итерируемый
объект, а **операторы** — только множества:

```python
print(small_set.issubset([1, 2, 3]))
print(big_set.issuperset([1, 2]))
```

```python
small_set <= [1, 2, 3]
```

В отличие от чисел, множества задают **частичный порядок**: два множества
могут быть **несравнимы** — ни одно из них не является подмножеством другого.
Тогда все сравнения `<`, `>` и `==` возвращают `False`. (Не путайте это со
сравнением списков и кортежей из занятия 4: там порядок лексикографический,
и любые два списка со сравнимыми элементами можно сравнить.)

```python
print(sorted(first_set), sorted(second_set))

print("first < second: ", first_set < second_set)
print("first > second: ", first_set > second_set)
print("first == second:", first_set == second_set)
```

## 4. Замороженные множества

**Замороженное множество** (`frozenset`) — это **неизменяемая** версия
множества. Создается функцией `frozenset(iterable)`; дубликаты, как и у
обычного множества, отбрасываются.

```python
frozen_numbers = frozenset([3, 1, 2, 3, 1])

print(frozen_numbers)
print(sorted(frozen_numbers))
print(2 in frozen_numbers)
```

Методов, изменяющих содержимое (`add()`, `remove()`, `update()` и т. д.), у `frozenset` нет:

```python
frozen_numbers.add(4)
```

### Зачем нужен `frozenset`

Обычное множество изменяемо, а значит **нехешируемо**: поместить его внутрь
другого множества или использовать как ключ словаря нельзя.

```python
{{1, 2}, {3, 4}}
```

`frozenset` неизменяем и поэтому **хешируем**. Его можно хранить во
множестве. Заодно видно, что порядок элементов не важен: `{1, 2}` и `{2, 1}`
— одно и то же множество.

```python
set_of_frozensets = {frozenset({1, 2}), frozenset({3, 4}), frozenset({2, 1})}

print(len(set_of_frozensets))
```

И использовать как ключ словаря. Например, расстояние между двумя городами не
зависит от порядка, в котором мы их назвали:

```python
distances = {
    frozenset({"Moscow", "Paris"}): 2490,
    frozenset({"Moscow", "Berlin"}): 1610,
}

print("Moscow - Paris:", distances[frozenset({"Moscow", "Paris"})])
print("Paris - Moscow:", distances[frozenset({"Paris", "Moscow"})])
```

### Операции над `frozenset`

Все операции теории множеств работают и для замороженных множеств,
результат — тоже `frozenset`:

```python
first_frozen = frozenset({1, 2, 3})
second_frozen = frozenset({3, 4})

print("Union:               ", first_frozen | second_frozen)
print("Intersection:        ", first_frozen & second_frozen)
print("Difference:          ", first_frozen - second_frozen)
print("Symmetric difference:", first_frozen ^ second_frozen)
```

Если смешать `set` и `frozenset`, тип результата определяется **левым
операндом**:

```python
print(type(first_frozen | {9}).__name__)
print(type({9} | first_frozen).__name__)
```

Сравнение `frozenset` и `set` учитывает только элементы:

```python
print(frozenset({1, 2}) == {1, 2})
```
