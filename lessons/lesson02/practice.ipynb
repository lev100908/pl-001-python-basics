## 1. Ввод и вывод данных: `input` и `print`

- `input(prompt)` — останавливает выполнение программы и ждет, пока
  пользователь введет строку и нажмет Enter. Введенный текст возвращается
  **всегда как `str`**, даже если пользователь ввел число.
- Если нужно число, строку нужно явно преобразовать: `int(...)` для целых,
  `float(...)` для дробных.
- `print(obj1, obj2, ..., objN, sep=" ", end="\n")` — выводит один или несколько объектов
  через пробел (по умолчанию) и выполняет перевод на новую строку.

````python
name = input("Как вас зовут? ")
print("Привет,", name)
````

````python
age_str = input("Сколько вам лет? ")
age = int(age_str)  # без преобразования age_str останется строкой

print("Через 10 лет вам будет", age + 10, "лет")
````

`print` можно настраивать через именованные параметры `sep` (разделитель

````python
print("a", "b", "c", sep=" -> ")
print("Без", end=" ")
print("переноса строки")
````

## 2. Цикл `while`: `break`, `continue`, `else`

- `break` — немедленно завершает цикл целиком, минуя оставшиеся итерации и
  проверку условия.
- `continue` — прерывает **текущую** итерацию и сразу переходит к проверке
  условия `while` (не завершая цикл).
- `else` у `while` — необязательный блок **после** цикла. Он выполняется,
  если цикл завершился штатно — потому что условие стало ложным. Если цикл
  прервали через `break`, блок `else` **не выполняется**.

Зачем это нужно на практике:

- `break` избавляет от лишней работы и лишних переменных. Например, при
  поиске элемента можно было бы завести флаг `found` и добавить его в
  условие `while`, но это усложняет код. С `break` достаточно выйти из
  цикла сразу, как только результат найден.
- `continue` помогает пропускать "неподходящие" элементы, не оборачивая
  весь остаток тела цикла во вложенный `if`. Тело цикла остается плоским
  и читается линейно, а не превращается в лесенку из вложенных условий.
- `else` полезен, когда важно отличить "цикл закончился сам" от "цикл
  прервали досрочно", например при поиске элемента — без него пришлось бы
  заводить отдельную переменную-флаг для этой проверки после цикла.

````python
# break: останавливаем цикл, как только нашли то, что искали
numbers = [4, 7, 2, 9, 5]
target = 9

index = 0
while index < len(numbers):
    if numbers[index] == target:
        print("Нашли", target, "на позиции", index)
        break

    index += 1
````

````python
# continue: пропускаем нечетные числа, печатаем только четные
current = 0
while current < 10:
    current += 1

    if current % 2 != 0:
        continue

    print(current)
````

````python
# while / else: else выполняется, если цикл не был прерван через break
numbers = [4, 7, 2, 9, 5]
target = 100  # такого числа в списке нет

index = 0
while index < len(numbers):
    if numbers[index] == target:
        print("Нашли", target)
        break

    index += 1

else:
    print(target, "в списке не найден")
````

### `break` и `continue` во вложенных циклах

Циклы можно вкладывать друг в друга — тогда на каждой итерации внешнего
цикла будет полностью отрабатывать внутренний цикл. Важно понимать: `break`
и `continue` относятся **только к тому циклу, в теле которого они
написаны** — то есть к ближайшему циклу, внутри которого находится строка с
`break`/`continue`. На выполнение внешнего цикла они никак не влияют — он
продолжает работать как обычно.

````python
# break во внутреннем цикле останавливает только его,
# внешний цикл как ни в чем не бывало переходит к следующей итерации
outer = 1
while outer <= 3:
    print("Внешний цикл, итерация", outer)

    inner = 1
    while inner <= 5:
        if inner == 3:
            break  # прерывает только внутренний while

        print("  внутренний цикл, итерация", inner)
        inner += 1

    outer += 1
````

````python
# continue во внутреннем цикле пропускает итерацию только внутреннего цикла,
# внешний цикл продолжает считать свои итерации как обычно
outer = 1
while outer <= 3:
    print("Внешний цикл, итерация", outer)

    inner = 1
    while inner <= 4:
        if inner % 2 == 0:
            inner += 1
            continue  # пропускает только шаг внутреннего while

        print("  внутренний цикл, итерация", inner)
        inner += 1

    outer += 1
````

## 4. Цикл `for` и объект `range`

`for` перебирает элементы **последовательности** (например, строки, списка
или диапазона чисел) один за другим и на каждой итерации присваивает
очередной элемент переменной цикла — не нужно вручную заводить и изменять
счетчик, как в `while`.

```python
for элемент in последовательность:
    тело_цикла
```

Чтобы просто повторить действие заданное число раз, используют `range`:

- `range(stop)` — числа от `0` до `stop - 1`;
- `range(start, stop)` — числа от `start` до `stop - 1`;
- `range(start, stop, step)` — то же самое, но с шагом `step`.

`range` не включает правую границу `stop`.

````python
for i in range(5):
    print(i)
````

````python
for i in range(2, 10, 2):
    print(i)
````

### `break`, `continue` и `else` в цикле `for`

`for` поддерживает `break`, `continue` и `else` точно так же, как и `while`:

- `break` — немедленно завершает цикл `for` целиком, минуя оставшиеся
  элементы последовательности.
- `continue` — прерывает **текущую** итерацию и переходит к **следующему
  элементу** последовательности (в `while` — к проверке условия).
- `else` у `for` — выполняется, если цикл перебрал всю последовательность
  и не был прерван через `break`.

Смысл использования тот же, что и для `while`: `break` избавляет от лишней
работы после того, как результат уже найден, `continue` позволяет
пропускать "неподходящие" элементы без вложенного `if`, а `else` помогает
отличить "перебрали всё" от "вышли досрочно".

````python
# break: останавливаем цикл, как только нашли то, что искали
numbers = [4, 7, 2, 9, 5]
target = 9

for index, number in enumerate(numbers):
    if number == target:
        print("Нашли", target, "на позиции", index)
        break
````

````python
# continue: пропускаем нечетные числа, печатаем только четные
for current in range(1, 11):
    if current % 2 != 0:
        continue

    print(current)
````

````python
# for / else: else выполняется, если цикл не был прерван через break
numbers = [4, 7, 2, 9, 5]
target = 100  # такого числа в списке нет

for number in numbers:
    if number == target:
        print("Нашли", target)
        break

else:
    print(target, "в списке не найден")
````

### `break` и `continue` во вложенных циклах `for`

Как и в случае с `while`, `break` и `continue` относятся **только к тому
циклу `for`, в теле которого они написаны** — то есть к ближайшему
вложенному циклу. На внешний цикл они не влияют — он продолжает работать
как обычно и переходит к следующей своей итерации.

Обратите внимание: с `for` про `continue` можно вообще не думать про
увеличение счетчика (в отличие от `while`, где перед `continue` нужно не
забыть его обновить) — `range` и так вернется на следующую итерацию с
правильным значением.

````python
# break во внутреннем цикле останавливает только его,
# внешний цикл как ни в чем не бывало переходит к следующей итерации
for outer in range(1, 4):
    print("Внешний цикл, итерация", outer)

    for inner in range(1, 6):
        if inner == 3:
            break  # прерывает только внутренний for

        print("  внутренний цикл, итерация", inner)
````

````python
# continue во внутреннем цикле пропускает итерацию только внутреннего цикла,
# внешний цикл продолжает считать свои итерации как обычно
for outer in range(1, 4):
    print("Внешний цикл, итерация", outer)

    for inner in range(1, 5):
        if inner % 2 == 0:
            continue  # пропускает только шаг внутреннего for

        print("  внутренний цикл, итерация", inner)
````
